```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_excel("/kaggle/input/datasets/ozfmomani/employee-data/Employee Sample Data - A(1).xlsx")
```

```python
print(f"Rows: {df.shape[0]}  |  Columns: {df.shape[1]}")
missing = df.isnull().sum()
print("\nMissing Values:")
print(pd.DataFrame({'Count': missing})[missing > 0])

print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

```python
cols_to_drop = ['Job Title', 'Department', 'Gender', 'Ethnicity', 'Country', 'City','Hire Date']
df = df.dropna(subset=cols_to_drop)
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Annual Salary'] = df['Annual Salary'].fillna(df['Annual Salary'].median())
df['Bonus %'] = df['Bonus %'].fillna(0)
print(f"Rows after cleaning: {df.shape[0]}")
print("\nMissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['Hire Year'] = df['Hire Date'].dt.year
hired_per_year = df[df['Exit Date'].isna()].groupby('Hire Year').size().reset_index(name='Count')

fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(hired_per_year['Hire Year'], hired_per_year['Count'],
        color='#2E75B6', linewidth=2.5, marker='o',
        markersize=7, markerfacecolor='white', markeredgewidth=2)

ax.fill_between(hired_per_year['Hire Year'], hired_per_year['Count'],
                alpha=0.15, color='#2E75B6')

for _, row in hired_per_year.iterrows():
    ax.text(row['Hire Year'], row['Count'] + 1.5,
            str(int(row['Count'])),
            ha='center', fontsize=9, fontweight='bold', color='#1F4E79')

ax.set_title('Number of Employees Hired Per Year', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Year')
ax.set_ylabel('Number of Employees Hired')
ax.spines[['top', 'right']].set_visible(False)
ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Gender & Age Distribution', fontsize=16, fontweight='bold')

BLUE_MID   = '#2E75B6'
BLUE_DARK  = '#1F4E79'
BLUE_LIGHT = '#9DC3E6'
GREEN      = '#70AD47'

gender = df[df['Exit Date'].isna()]['Gender'].value_counts()

axes[0].pie(gender.values, labels=gender.index, autopct='%1.1f%%',
            colors=[BLUE_DARK, BLUE_LIGHT], startangle=90)
axes[0].set_title('Male vs. Female')

bins   = [0, 30, 40, 50, 65]
labels = ['Under 30', '30–39', '40–49', '50+']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
age_group = df[df['Exit Date'].isna()]['Age Group'].value_counts().sort_index()

axes[1].bar(age_group.index, age_group.values,
            color=[BLUE_LIGHT, BLUE_MID, BLUE_DARK, GREEN])
axes[1].set_title('Employees by Age Group')
axes[1].set_xlabel('Age Group')
axes[1].set_ylabel('Number of Employees')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Geographic & Diversity Distribution', fontsize=16, fontweight='bold')

BLUE_DARK  = '#1F4E79'
BLUE_MID   = '#2E75B6'
BLUE_LIGHT = '#9DC3E6'
GREEN      = '#70AD47'

country = df[df['Exit Date'].isna()]['Country'].value_counts().sort_values(ascending=True)
axes[0].barh(country.index, country.values,
             color=[BLUE_LIGHT, BLUE_MID, BLUE_DARK], edgecolor='white')
axes[0].set_title('Employees by Country')
axes[0].set_xlabel('Number of Employees')

city = df[df['Exit Date'].isna()]['City'].value_counts().sort_values(ascending=True)
axes[1].barh(city.index, city.values, color=BLUE_MID, edgecolor='white')
axes[1].set_title('Employees by City')
axes[1].set_xlabel('Number of Employees')

eth = df[df['Exit Date'].isna()]['Ethnicity'].value_counts().sort_values(ascending=True)
axes[2].barh(eth.index, eth.values,
             color=[BLUE_LIGHT, BLUE_MID, BLUE_DARK, GREEN], edgecolor='white')
axes[2].set_title('Employees by Ethnicity')
axes[2].set_xlabel('Number of Employees')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Department & Business Unit Analysis', fontsize=16, fontweight='bold')

BLUE_DARK  = '#1F4E79'
BLUE_MID   = '#2E75B6'
BLUE_LIGHT = '#9DC3E6'
GREEN      = '#70AD47'

dept_counts = df[df['Exit Date'].isna()]['Department'].value_counts().sort_values(ascending=True)
axes[0,0].barh(dept_counts.index, dept_counts.values, color=BLUE_MID, edgecolor='white')
axes[0,0].set_title('Num of Employees by Department')
axes[0,0].set_xlabel('Number of Employees')

avg_sal_dept = df[df['Exit Date'].isna()].groupby('Department')['Annual Salary'].mean().sort_values(ascending=True)
axes[0,1].barh(avg_sal_dept.index, avg_sal_dept.values, color=BLUE_DARK, edgecolor='white')
axes[0,1].set_title('Avg Annual Salary by Department')
axes[0,1].set_xlabel('Avg Annual Salary (USD)')

bu_counts = df[df['Exit Date'].isna()]['Business Unit'].value_counts().sort_values(ascending=True)
axes[1,0].barh(bu_counts.index, bu_counts.values, color=GREEN, edgecolor='white')
axes[1,0].set_title('Num of Employees by Business Unit')
axes[1,0].set_xlabel('Number of Employees')

avg_sal_bu = df[df['Exit Date'].isna()].groupby('Business Unit')['Annual Salary'].mean().sort_values(ascending=True)
axes[1,1].barh(avg_sal_bu.index, avg_sal_bu.values, color=GREEN, edgecolor='white')
axes[1,1].set_title('Avg Annual Salary by Business Unit')
axes[1,1].set_xlabel('Avg Annual Salary (USD)')

plt.tight_layout()
plt.show()
```
